Task 1: Perform Basic Mathematical Operations

Problem Statement: Write a Python program that does the following:

Takes two numbers as input from the user.

Performs the basic mathematical operations on these two numbers:

Addition
Subtraction
Multiplication
Division
Displays the results of each operation on the screen.

```python
"""We will use the input() function, to take the two numbers as input from the user."""

num1=input("Enter the first number:")    # input(), prompts the user to enter the numbers.
num2=input("Enter the second number:")

"""Basic Mathematical operations."""

print('Addition:',int(num1)+int(num2)) # addition
print('Subtraction:',int(num1)-int(num2)) #subtraction
print('Multiplication:',int(num1)*int(num2)) #multiplication
print('Division:',float(num1)/float(num2)) #division
```
